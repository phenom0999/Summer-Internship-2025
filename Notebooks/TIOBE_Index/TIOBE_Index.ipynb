```python
import pandas as pd

df = pd.read_csv("data/Tiobe_Index_All_Ratings_January2026.csv")
```

```python
# convert DATE column to datetime
df["DATE"] = pd.to_datetime(df["DATE"], format="%b %d, %Y")

# keep only rows from 2018
df_2018 = df[df["DATE"].dt.year >= 2018]
```

```python
import matplotlib.pyplot as plt
import pandas as pd

top_cols = ["Java", "Python", "C", "C++", "C#"]

ax = df.set_index("DATE")[top_cols].plot(figsize=(12, 6))

# ChatGPT-3.5 release date
gpt35_release = pd.Timestamp("2022-11-30")

ax.axvline(
    gpt35_release,
    linestyle="--",
    linewidth=1,
    label="ChatGPT-3.5 Release",
    color="brown"
)

plt.xlabel("Year")
plt.ylabel("Index value")
plt.title("LLM impact on the TIOBE Index")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.savefig("../../Literature_Review/images/impact_on_TIOBE_index/tiobe_index.png")
plt.show()

```
